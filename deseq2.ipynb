```r
library(DESeq2)
regions <- c("anterior_cingulate", "caudate", "frontal_cortex", "nucleus_accumbens", "putamen")
dir.create("deseq_results", showWarnings = FALSE)
```

```r
run_region <- function(region) {
  counts  <- read.csv(paste0("region_data/", region, "_counts.csv"),  row.names = 1, check.names = FALSE)
  samples <- read.csv(paste0("region_data/", region, "_samples.csv"), row.names = 1, check.names = FALSE)
  counts  <- round(as.matrix(counts[, rownames(samples)]))

  samples$sex      <- factor(samples$sex, levels = c("female", "male"))
  samples$AGE      <- factor(samples$AGE)
  samples$RIN      <- as.numeric(scale(samples$SMRIN))
  samples$ischemic <- as.numeric(scale(samples$SMTSISCH))

  dds <- DESeqDataSetFromMatrix(counts, samples, design = ~ AGE + RIN + ischemic + sex)
  dds <- DESeq(dds)
  res <- results(dds, contrast = c("sex", "male", "female"), alpha = 0.05)

  out <- as.data.frame(res)
  out$gene_id <- rownames(out)
  write.csv(out, paste0("deseq_results/", region, "_sex_DE.csv"), row.names = FALSE)

  cat(region, ": male-biased =", sum(res$padj < 0.05 & res$log2FoldChange > 0, na.rm = TRUE),
      "| female-biased =",       sum(res$padj < 0.05 & res$log2FoldChange < 0, na.rm = TRUE), "\n")
}

system.time(run_region("anterior_cingulate"))
```

```r
for (region in regions[-1]) run_region(region)
```
